```python
### Usemos un diccionario para almacenar la nómina de los empleados
empleados = {
    "Alice" : 1_000_000,
    "Bob" : 998_671,
    "Carol" : 122_908,
    "Frank" : 88_123, 
    "Eve" : 93_121
}
## Genera un nuevo diccionario que contenga sólo los empleados que ganen arriba de 100_000
nuevo_diccionario = {empleado : salario for empleado, salario in empleados.items() if salario > 100_000}
nuevo_diccionario

## Genera una lista que contenga los nombres de los empleados que ganen arriba de 100_00
nombre_empleados = [empleado for empleado, salario in empleados.items() if salario > 100_000]
nombre_empleados
```

```python
#### Genera un diccionario que contenga la compensación del 10% del salario para cada empleado
tasa_compensacion = 0.10
{empleado : salario + round(salario*tasa_compensacion,1)  for empleado, salario in empleados.items() if salario > 100_000}
```

```python
def compensacion_total(salario, tasa, precision):
    return round(salario+(salario*tasa),precision)

{empleado : compensacion_total(salario, tasa_compensacion, 1) for empleado, salario in empleados.items() if salario > 100_000}
```
