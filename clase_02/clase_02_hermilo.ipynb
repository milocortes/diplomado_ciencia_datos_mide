```python
from typing import Union
### Tipe hint

def modifica_lista(la_lista : list, valor : Union[int, float, bool]) -> None:
    """
    ## Documentacion
    La funcion agrega un valor al final de la lista.

    Argumentos:

    - la_lista :  list  ---> Es la lista a modificar
    - valor : Union[int, float, bool] ---> Valor a agregar a la lista

    Valor de regreso None
    """
    la_lista.append(valor)

b = [2,3,4,5]

k = modifica_lista(b, 102)
```
